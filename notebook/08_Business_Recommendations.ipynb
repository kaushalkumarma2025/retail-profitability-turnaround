# Step 5: Business Recommendations

## 5.1 Objective

The objective is to translate the profitability analysis into actionable business recommendations for improving retail profitability. The analysis identified two loss-making subcategories—Tables and Bookcases—with different levels of profitability deterioration across products, discount levels, and regions.

The recommendations prioritize the most severe observed losses while accounting for the limitations of historical, observational data.

## 5.2 Recommendation 1: Review Discounting Practices for Tables in the East Region

**Evidence**

- Net loss: **$11,025.38**
- Profit margin: **−28.17%**
- Loss-making records: **97.50%**
- Average discount: **37%**

Tables in the East region represent the most severe profitability issue identified in the analysis. The combination of a deeply negative profit margin, a high average discount, and a very high share of loss-making records makes this segment the first priority for investigation.

**Recommended action**

Review high-discount Table transactions in the East region to identify products and order characteristics associated with negative profitability. Introduce a proposed margin-based discount approval process that flags orders expected to generate negative profit for review before approval.

Discount limits should be determined using product-level cost and margin information rather than applying an arbitrary uniform cap.

**KPIs to monitor**

- Table profit margin in the East region
- Net profit from Table transactions
- Share of loss-making Table records
- Average discount on Table transactions
- Sales volume and order conversion following pricing changes

**Expected business rationale**

Reducing unprofitable discounting may improve margins, provided that any reduction in demand does not offset the gains. The actual impact must be measured through a controlled pilot or a suitable before-and-after evaluation.

## 5.3 Recommendation 2: Investigate Persistently Loss-Making Products

**Evidence**

The five products with the largest negative total profits accounted for approximately **29.1% of the negative profit summed across all loss-making products** in Tables and Bookcases.

The largest contributors included:

- Chromcraft Bull-Nose Wood Oval Conference Table: **−$2,876.12**
- Bush Advantage Collection Racetrack Conference Table: **−$1,934.40**
- Balt Solid Wood Round Tables: **−$1,201.06**

These products are candidates for targeted investigation because their aggregate profitability was particularly poor in the observed dataset.

**Recommended action**

Conduct a product-level profitability review covering:

1. Selling prices and discount levels.
2. Product acquisition costs and other relevant costs.
3. Order quantities and transaction characteristics.
4. Profitability across regions and customer segments.

Based on the findings, management could consider product-specific pricing changes, tighter discount approvals, supplier-cost negotiations, or changes to product assortment where commercially justified.

**KPIs to monitor**

- Net profit by product
- Product-level profit margin
- Frequency of loss-making transactions
- Average discount by product
- Sales volume and profitability after any intervention

**Expected business rationale**

Targeted interventions may address a meaningful portion of the observed negative profitability without imposing unnecessary restrictions across the entire product portfolio.

## 5.4 Recommendation 3: Investigate Bookcases Profitability in the Central Region

**Evidence**

- Net loss: **$1,997.90**
- Profit margin: **−8.27%**
- Loss-making records: **72.00%**

Bookcases in the Central region had the weakest regional performance for this subcategory. In comparison, Bookcases in the South generated a positive profit of $1,339.49 and a profit margin of 12.29%.

This regional variation suggests an opportunity to investigate differences in product mix, discounts, costs, and order characteristics.

**Recommended action**

Compare Bookcases transactions in the Central region with those in the South. Identify whether differences in discount levels, product selection, transaction sizes, or costs help explain the profitability gap.

Use the results to design targeted changes for Central rather than assuming that a single pricing policy will work equally well across regions.

**KPIs to monitor**

- Bookcases profit margin by region
- Net profit by region
- Loss-making transaction share
- Average discount by region
- Product mix and sales volume

**Expected business rationale**

A region-specific investigation may identify commercially actionable differences and help management focus on the factors most closely associated with poor profitability.

## 5.5 Recommendation 4: Establish Ongoing Profitability Monitoring

**Evidence**

Across Tables and Bookcases, the analysis identified:

- Combined sales of **$321,845.53**
- Combined net loss of **$21,198.04**
- Overall profit margin of **−6.59%**
- **312 loss-making records out of 547**, equivalent to 57.04%

The analysis also identified substantial variation in profitability across discount levels, products, and regions.

**Recommended action**

Develop a recurring profitability dashboard or reporting process that tracks performance at the product, subcategory, and regional levels. Flag transactions with negative expected margins and require review where appropriate.

The monitoring process should distinguish between individual loss-making transactions and aggregate net profitability, because profitable transactions can offset losses elsewhere.

**KPIs to monitor**

- Net profit and profit margin
- Loss-making transaction share
- Average discount
- Profitability by product and region
- Sales volume and order conversion
- Changes in profitability following pricing interventions

**Expected business rationale**

Regular monitoring can help management identify deteriorating margins earlier, prioritize investigation, and assess whether pricing or product-level interventions are improving financial performance.

## 5.6 Implementation Priorities

| Priority | Proposed intervention | Evidence supporting priority |
|---|---|---|
| 1 | Review Table discounting in the East region | Largest regional loss and lowest regional margin |
| 2 | Investigate the worst-performing Table products | Largest individual product-level negative profits |
| 3 | Investigate Bookcases in the Central region | Weakest regional Bookcases margin |
| 4 | Establish recurring profitability monitoring | 57.04% of target records are loss-making |

These priorities reflect the severity of observed profitability problems. They do not imply that the first intervention will necessarily produce the greatest recoverable financial benefit.

## 5.7 Risks and Limitations

The recommendations are based on historical observational data and should be treated as hypotheses for business action rather than proven solutions.

- **Association is not causation:** Higher discounts coincide with poorer profitability in several groups, but the analysis does not establish that discounts alone caused the losses.
- **Demand response is unknown:** Reducing discounts could lower sales volume or order conversion.
- **Cost information is limited:** Further product-cost analysis may be required to distinguish pricing issues from cost or product-mix effects.
- **Historical losses are not recoverable savings:** The observed negative profit cannot be assumed to be fully recoverable through pricing changes.
- **Regional comparisons may reflect different transaction mixes:** Differences across regions should be investigated before transferring practices from a profitable region to an unprofitable one.

## 5.8 Conclusion

The analysis identifies Tables as the primary profitability concern, particularly in the East region, while Bookcases in the Central region represent a secondary area for investigation. Product-level results also indicate that a subset of Table products contributes disproportionately to the negative profit summed across loss-making products.

The recommended approach is to prioritize the most severe segments, investigate product economics and discount practices, introduce margin-based transaction reviews, and monitor outcomes after any intervention.

The next stage of business evaluation should test whether these interventions improve net profitability without causing an offsetting decline in sales volume. No financial improvement is claimed until it has been measured.
