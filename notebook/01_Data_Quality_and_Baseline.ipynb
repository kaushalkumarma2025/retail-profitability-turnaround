```python
import pandas as pd

df = pd.read_csv("../data/Sample - Superstore.csv", encoding="cp1252")

print("Dataset loaded successfully")
print("Shape:", df.shape)

display(df.head())
```

```python
print("Columns:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Convert date columns to datetime
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# Check date conversion and chronology
print("Invalid order dates:", df["Order Date"].isna().sum())
print("Invalid ship dates:", df["Ship Date"].isna().sum())

print("Earliest order:", df["Order Date"].min())
print("Latest order:", df["Order Date"].max())

print(
    "Orders shipped before order date:",
    (df["Ship Date"] < df["Order Date"]).sum()
)

# Check numeric ranges
print("\nNumeric summary:")
display(df[["Sales", "Quantity", "Discount", "Profit"]].describe())

print("\nNegative sales:", (df["Sales"] < 0).sum())
print("Zero sales:", (df["Sales"] == 0).sum())
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Negative discounts:", (df["Discount"] < 0).sum())
print("Discounts above 100%:", (df["Discount"] > 1).sum())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
profit_margin = total_profit / total_sales * 100

print(f"Total sales: ${total_sales:,.2f}")
print(f"Total profit: ${total_profit:,.2f}")
print(f"Total quantity sold: {total_quantity:,}")
print(f"Overall profit margin: {profit_margin:.2f}%")

print(
    "Loss-making rows:",
    (df["Profit"] < 0).sum()
)

print(
    "Share of loss-making rows:",
    f"{(df['Profit'] < 0).mean() * 100:.2f}%"
)
```
